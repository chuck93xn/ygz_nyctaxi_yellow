```python
import os
import sys
from pathlib import Path

def setup_project_path():
    try:
        project_root = Path(__file__).resolve().parents[2]
        if str(project_root) not in sys.path:
            sys.path.append(str(project_root))
        return project_root
    except NameError:

        return None
PROJECT_ROOT = setup_project_path()
```

```python
from pyspark.sql.functions import date_format
from modules.utils.date_utils import get_month_start_n_months_ago
```

```python
start_date = get_month_start_n_months_ago(2)
df = spark.read.table("nyctaxi_ygz.`02_silver`.yellow_trips_enriched").filter(f"pickup_datetime > '{start_date}'")
```

```python
df = df.withColumn("year_month", date_format("pickup_datetime", "yyyy-MM"))
```

```python
df.write.\
    option("path", "abfss://nyctaxi-dbw@nyctaxidbwexport.dfs.core.windows.net/Ygz_yellow_trips_export/").\
    format("json").\
    mode("append").\
    partitionBy("vendor", "year_month").\
    saveAsTable("nyctaxi_ygz.04_export.yellow_trips_export")
```
