```python
from pyspark.sql.functions import date_format
```

```python
df = spark.read.table("nyctaxi_ygz.`02_silver`.yellow_trips_enriched")
```

```python
df = df.withColumn("year_month", date_format("pickup_datetime", "yyyy-MM"))
```

```python
df.write.\
    option("path", "abfss://nyctaxi-dbw@nyctaxidbwexport.dfs.core.windows.net/Ygz_yellow_trips_export/").\
    format("json").\
    mode("overwrite").\
    partitionBy("vendor", "year_month").\
    saveAsTable("nyctaxi_ygz.04_export.yellow_trips_export")
```
